```python
import pandas as pd

emissions = pd.read_csv("https://koumoul.com/s/data-fair/api/v1/datasets/igt-pouvoir-de-rechauffement-global/convert")
emissions.head(2)
```

```python
emissions_copy=emissions.loc[:,["INSEE commune", "Commune", "Autres transports", "Autres transports international"]]
print(emissions_copy)

emissions_copy=emissions_copy.rename(columns={"INSEE commune": "code_insee", "Autres transports": "transports", "Autres transports international":"transport_international"})

emissions_copy=emissions_copy.fillna(0)
#emissions_copy["dep"]=emissions_copy["code_insee"].str[:2]
```

```python
emissions_copy["dep"]=emissions_copy["code_insee"].str[:2]
emissions_copy["transports_total"]=emissions_copy["transports"]+emissions_copy["transport_international"]
emissions_copy.head(5)
```

```python
emissions_copy_tri=emissions_copy.sort_values(by="transports_total", ascending=False)

emissions_copy_sum=emissions_copy.groupby(by="dep")["transports_total"].sum().reset_index(name="transports_total_par_dep")
emissions_copy_sum.head(10)
emissions_copy_sum_tri =emissions_copy_sum.sort_values(by="transports_total_par_dep", ascending=False)
emissions_copy_sum_tri.head(10)
```

```python
emissions_copy_select = emissions_copy.loc[emissions_copy["dep"].isin(["13", "31"])]
emissions_copy_select.head(10)

emissions_copy_select_tri=emissions_copy_select.sort_values(by="transports_total", ascending=False)
emissions_copy_select_tri.head(10)
```
